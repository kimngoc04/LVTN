```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os

def generate_pure_explicit_87_minutes():
    # Khai báo tường minh đầy đủ 87 phút mô phỏng theo Hình 5b của bài báo
    explicit_data = {
        1: 6789, 2: 5817, 3: 5607, 4: 8310, 5: 10049, 6: 10918, 7: 11363, 8: 10680, 9: 7800, 10: 6407,
        11: 10381, 12: 14750, 13: 11489, 14: 11994, 15: 13651, 16: 11942, 17: 9888, 18: 9989, 19: 17075, 20: 19176,
        21: 15004, 22: 20407, 23: 15354, 24: 14564, 25: 21297, 26: 25125, 27: 25564, 28: 27047, 29: 29136, 30: 22247,
        31: 20472, 32: 32722, 33: 42363, 34: 47217, 35: 50122, 36: 48945, 37: 35103, 38: 34909, 39: 80298, 40: 119650,
        41: 274217, 42: 310218, 43: 290332, 44: 162119, 45: 131450, 46: 317542, 47: 255797, 48: 228249, 49: 214978, 50: 200683,
        51: 117487, 52: 105441, 53: 203415, 54: 278085, 55: 289494, 56: 299658, 57: 393360, 58: 167530, 59: 121040, 60: 332333,
        61: 423227, 62: 205785, 63: 115387, 64: 241679, 65: 129041, 66: 81302, 67: 98828, 68: 229872, 69: 218168, 70: 122831,
        71: 74069, 72: 56740, 73: 78575, 74: 125558, 75: 64648, 76: 40346, 77: 30654, 78: 25120, 79: 16708, 80: 12429,
        81: 16352, 82: 16413, 83: 14735, 84: 13431, 85: 12052, 86: 9228, 87: 7831,
    }
    
    t_full = np.arange(1, 88)
    base_workload = np.array([explicit_data[m] for m in t_full])
    return t_full, base_workload

# Thực thi lấy dữ liệu
minutes, workload_values = generate_pure_explicit_87_minutes()

df = pd.DataFrame({
    'time': minutes,
    'requests': workload_values
})

# Đảm bảo thư mục tồn tại trước khi lưu file
os.makedirs('../data', exist_ok=True)
csv_filename = '../data/worldcup_workload.csv'
df.to_csv(csv_filename, index=False)
print(f"✅ Đã xuất thành công file '{csv_filename}' với đúng {len(df)} dòng dữ liệu World Cup!")

print(df.head(5))
print(df.tail(5))

# Vẽ và lưu biểu đồ kiểm tra
os.makedirs('../../workload_generation/data', exist_ok=True)
plt.figure(figsize=(6, 4.5))
plt.plot(df['time'], df['requests'], color='#0047AB', linewidth=1.2)
plt.title("World Cup", fontsize=12, fontweight='bold')
plt.xlabel("Time (min)", fontsize=10)
plt.ylabel("Total requests", fontsize=10)

plt.xlim(1, 90)
plt.ylim(0, 450000)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()

plt.savefig('../../workload_generation/data/worldcup_workload.png', dpi=300)
print("✅ Đã lưu biểu đồ World Cup thành công!")
plt.show()
```
