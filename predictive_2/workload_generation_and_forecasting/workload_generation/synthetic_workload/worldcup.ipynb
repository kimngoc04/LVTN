```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os

def generate_pure_explicit_87_minutes():
    # Khai báo tường minh đầy đủ 87 phút mô phỏng theo Hình 5b của bài báo
    explicit_data = {
        1: 678, 2: 581, 3: 560, 4: 831, 5: 1004, 6: 1091, 7: 1136, 8: 1068, 9: 780, 10: 640,
        11: 1038, 12: 1475, 13: 1148, 14: 1199, 15: 1365, 16: 1194, 17: 988, 18: 998, 19: 1707, 20: 1917,
        21: 1500, 22: 2040, 23: 1535, 24: 1456, 25: 2129, 26: 2512, 27: 2556, 28: 2704, 29: 2913, 30: 2224,
        31: 2047, 32: 3272, 33: 4236, 34: 4721, 35: 5012, 36: 4894, 37: 3510, 38: 3490, 39: 8029, 40: 11965,
        41: 27421, 42: 31021, 43: 29033, 44: 16211, 45: 13145, 46: 31754, 47: 25579, 48: 22824, 49: 21497, 50: 20068,
        51: 11748, 52: 10544, 53: 20341, 54: 27808, 55: 28949, 56: 29965, 57: 39336, 58: 16753, 59: 12104, 60: 33233,
        61: 42322, 62: 20578, 63: 11538, 64: 24167, 65: 12904, 66: 8130, 67: 9882, 68: 22987, 69: 21816, 70: 12283,
        71: 7406, 72: 5674, 73: 7857, 74: 12555, 75: 6464, 76: 4034, 77: 3065, 78: 2512, 79: 1670, 80: 1242,
        81: 1635, 82: 1641, 83: 1473, 84: 1343, 85: 1205, 86: 922, 87: 783,
    }
    
    t_full = np.arange(1, 88)
    base_workload = np.array([explicit_data[m] for m in t_full])
    return t_full, base_workload

# Thực thi lấy dữ liệu
minutes, workload_values = generate_pure_explicit_87_minutes()

df = pd.DataFrame({
    'time': minutes,
    'requests': workload_values
})

# Đảm bảo thư mục tồn tại trước khi lưu file
os.makedirs('../data', exist_ok=True)
csv_filename = '../data/worldcup_workload.csv'
df.to_csv(csv_filename, index=False)
print(f"✅ Đã xuất thành công file '{csv_filename}' với đúng {len(df)} dòng dữ liệu World Cup!")

print(df.head(5))
print(df.tail(5))

# Vẽ và lưu biểu đồ kiểm tra
os.makedirs('../../workload_generation/data', exist_ok=True)
plt.figure(figsize=(6, 4.5))
plt.plot(df['time'], df['requests'], color='#0047AB', linewidth=1.2)
plt.title("World Cup", fontsize=12, fontweight='bold')
plt.xlabel("Time (min)", fontsize=10)
plt.ylabel("Total requests", fontsize=10)

plt.xlim(1, 90)
plt.ylim(0, 45000)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()

plt.savefig('../../workload_generation/data/worldcup_workload.png', dpi=300)
print("✅ Đã lưu biểu đồ World Cup thành công!")
plt.show()
```
